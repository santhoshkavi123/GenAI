### 1. Read the PDF File 

```python
def DocumentViewer(hr_policy_docs:List[str], doc_page:int) -> None:
    print("=== Viewing: Page Content === \n")
    display(Markdown(hr_policy_docs[doc_page].page_content))
    print("=== Viewing: Metadata Information === \n")
    print( hr_policy_docs[doc_page].metadata)
```

```python
input_file_path = config['inputs']['document_input_file_path']
loader = PyMuPDFLoader(
    file_path = input_file_path, 
    mode = 'page', 
    extract_images = False
)

hr_policy_documents = loader.load()
print(f"len of HR policy documents: {len(hr_policy_documents)}")
```

```python
DocumentViewer(hr_policy_documents, 11)
```
